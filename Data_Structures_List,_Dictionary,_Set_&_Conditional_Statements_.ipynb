# Data Structures - List, Dictionary, Set & Conditional Statements

# **LIST**

```python
# List Creation:
age_list = [24,25,27,28,29]
name_list = ["saba", "jiilu", "asmi","suji", "dojil"]

# Append the string "Yazhini" to name_list
name_list.append("yazhini")
print(name_list)
```

```python
#  Insert the element 26 at index 2 in age_list

age_list = [24,25,27,28,29]

age_list.insert(2, 26)
print(age_list)
```

```python
# Remove the string "Yazhini" from name_list
name_list.remove("yazhini")
print(name_list)
```

```python
# Pop the last element from age_list
age_list.pop()
print(age_list)
```

```python
# Extend the age_list with additional ages [31,30,32]
age_list.extend([31,32,33])
print(age_list)
```

```python
 # Sort age_list in descending order
age_list = [24,25,26,27,28,29,31,32,33]
age_list.sort()
print(age_list)
```

```python
# Find the max age, min age, and sum of all ages from age_list
# max
max_age = max(age_list)
print(max_age)

#min
min_age = min(age_list)
print(min_age)

#sum
sum_age = sum(age_list)
print(sum_age)
```

```python
# slicing
# Print the first element of name_list
print(name_list[0])
```

```python
 # Print the last element of name_list
 print(name_list[-1])
```

```python
 # Print the elements from index 2 to index 4 in name_list
 print(name_list[2:5])
```

```python
# Print the elements of name_list in reverse order
name_list[::-1]
```

#**DICTIONARY**

```python
# Create a dictionary named student_marks that maps the names of five students
student_marks = {
    "sabana": 85,
    "jillu": 90,
    "asmi": 78,
    "suji": 92,
    "dojil": 88
}

print(student_marks)
```

```python
# print the mark of a asmi

student_marks["asmi"]
```

```python
#  Add a new student, "Janani," with a mark of 80 to the student_marks dictionary

student_marks["janani"] = 80
print(student_marks)
```

```python
#  Update the mark of any one older student to 82
student_marks["asmi"] = 82
print(student_marks)
```

```python
# Use the keys(), values(), and items() methods
print(student_marks.keys())
print(student_marks.values())
print(student_marks.items())
```

# **SET**

```python
#  Create a set

my_set = {'a','e','i','o','u','a','a','i'}

# Attempt to access my_set[4]. If the code throws an error, provide an explanation
my_set[4]

# A set has no index, so my_set[4] cannot be used to access an element
```

```python
# Create two sets:
set_1 = {1, 3, 5, 7, 9}
set_2 = {2, 3, 5, 8, 10}

# Compute and print the union and intersection of set_1 and set_2

#union
set_1.union(set_2)
```

```python
# intersection
set_1.intersection(set_2)
```
